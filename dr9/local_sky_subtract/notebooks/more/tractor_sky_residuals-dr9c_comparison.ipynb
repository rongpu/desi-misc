```python
tractor_file_list = []
tractor_dir_list = []

tractor_dir = '000'
tractor_file_list += glob.glob('/global/cscratch1/sd/ziyaoz/dr9c/tractor/'+tractor_dir+'/*.fits')
tractor_dir_list += [tractor_dir]*(len(tractor_file_list)-len(tractor_dir_list))

tractor_dir = '001'
tractor_file_list += glob.glob('/global/cscratch1/sd/ziyaoz/dr9c/tractor/'+tractor_dir+'/*.fits')
tractor_dir_list += [tractor_dir]*(len(tractor_file_list)-len(tractor_dir_list))

tractor_dir = '002'
tractor_file_list += glob.glob('/global/cscratch1/sd/ziyaoz/dr9c/tractor/'+tractor_dir+'/*.fits')
tractor_dir_list += [tractor_dir]*(len(tractor_file_list)-len(tractor_dir_list))

# print(tractor_file_list)
```

```python
# dr9c
cat_stack = []
for index in range(len(tractor_file_list)):
    cat = fitsio.read(tractor_file_list[index], columns=['ra', 'dec', 'brick_primary'])
    cat = Table(cat)
    mask = cat['brick_primary']==True
    # print(len(cat), np.sum(mask))
    cat = cat[mask]
    cat_stack.append(cat)
cat_stack = vstack(cat_stack)
print(len(cat_stack))
cat = cat_stack

plt.plot(cat['ra'][::5], cat['dec'][::5], '.', ms=0.2);
```

```python
# dr9c
cat_stack = []
for index in range(len(tractor_file_list)):
    cat = fitsio.read(tractor_file_list[index])
    cat = Table(cat)
    mask = cat['brick_primary']==True
    # print(len(cat), np.sum(mask))
    cat = cat[mask]
    cat_stack.append(cat)
cat_stack = vstack(cat_stack)
print(len(cat_stack))
cat = cat_stack
# dr8
cat_stack1 = []
for index in range(len(tractor_file_list)):
    cat1 = fitsio.read(os.path.join('/global/project/projectdirs/cosmo/data/legacysurvey/dr8/south/tractor/', tractor_dir_list[index], os.path.basename(tractor_file_list[index])))
    cat1 = Table(cat1)
    mask = cat1['brick_primary']==True
    # print(len(cat1), np.sum(mask))
    cat1 = cat1[mask]
    cat_stack1.append(cat1)
cat_stack1 = vstack(cat_stack1)
print(len(cat_stack1))
cat1 = cat_stack1
```

```python
# only look at PSF and REX objects
mask = (cat['type']=='PSF') | (cat['type']=='REX')
print(np.sum(mask)/len(mask))
cat = cat[mask]

mask = (cat1['type']=='PSF ') | (cat1['type']=='REX ')
print(np.sum(mask)/len(mask))
cat1 = cat1[mask]
```

```python
bands = ['g', 'r', 'z']
for band in bands: 
    # residual flux per sq arcsec
    cat[band+'_sky'] = (cat['apflux_resid_'+band][:, -1]-cat['apflux_resid_'+band][:, -2]) / (np.pi*7**2-np.pi*5**2)
    # residual flux per sq arcsec
    cat1[band+'_sky'] = (cat1['apflux_resid_'+band][:, -1]-cat1['apflux_resid_'+band][:, -2]) / (np.pi*7**2-np.pi*5**2)

# mask_clean = np.ones(len(cat), dtype=bool)
mask_clean = cat['maskbits']==0
mask_clean1 = cat1['maskbits']==0
```

```python
nmad = lambda x: 1.48 * np.median(np.abs(x-np.median(x)))

vranges = [0.01, 0.015, 0.03]
for band, vrange in zip(bands, vranges):

    plt.figure(figsize=(7, 5))
    tmp = cat[band+'_sky'][mask_clean]
    print('dr9c: {}-band sky residuals (nmgy/sq arcsec): median={:.5f}  NMAD={:.5f}'.format(band, np.median(tmp), nmad(tmp)))
    plt.hist(tmp, 100, range=(-vrange, vrange), density=True, alpha=0.5, label='dr9c: {:.5f}'.format(np.median(tmp)), color='C1')
    plt.axvline(np.median(tmp), color='C1', ls='--', lw=1)

    tmp = cat1[band+'_sky'][mask_clean1]
    print('DR8: {}-band sky residuals (nmgy/sq arcsec): median={:.5f}  NMAD={:.5f}'.format(band, np.median(tmp), nmad(tmp)))
    plt.hist(tmp, 100, range=(-vrange, vrange), density=True, alpha=0.5, label='dr8: {:.5f}'.format(np.median(tmp)), color='C0')
    plt.axvline(np.median(tmp), color='C0', ls='--', lw=1)
    # correct for the offset
    cat1[band+'_sky_corr'] = cat1[band+'_sky'] - np.median(tmp)

    plt.xlim(-vrange, vrange)
    plt.grid(alpha=0.5)
    plt.xlabel(band+'-band sky residuals (nmgy/sq arcsec) (masked)')
    plt.legend()
    plt.show()
```

```python
ms, alpha = 0.3, 1.

mask = np.ones(len(cat), dtype=bool)
mask1 = np.ones(len(cat1), dtype=bool)
# mask = (cat['dec']<0) & (cat['dec']>-1)
# mask1 = (cat1['dec']<0) & (cat1['dec']>-1)

ramax, ramin, decmin, decmax = 2.5, 0, -1.7, 0.5
# ramax, ramin, decmin, decmax = np.max(cat['ra'][mask]), np.min(cat['ra'][mask]),  np.min(cat['dec'][mask]), np.max(cat['dec'][mask])
xsize, ysize = 18, 18*((decmax-decmin)/(ramax-ramin)*np.cos((decmax+decmin)/2/180*np.pi))

for band, vrange in zip(bands, vranges):
 
    fig, ax = plt.subplots(figsize=(xsize, ysize))
    im = ax.scatter(cat['ra'][mask], cat['dec'][mask], c=cat[band+'_sky'][mask], vmin=-vrange, vmax=vrange, s=ms, alpha=alpha, cmap='seismic')
    ax.axis([ramax, ramin, decmin, decmax])
    ax.grid(alpha=0.5)
    ax.set_xlabel('ra')
    ax.set_ylabel('dec')
    # fig.colorbar(im)
    ax.set_title('[dr9c] '+band+'-band sky residuals [not masked] (color range: +-{})'.format(vrange))
    plt.savefig('plots/dr9c_'+band+'_sky_residuals_region1.png')
    plt.show()

    fig, ax = plt.subplots(figsize=(xsize, ysize))
    im = ax.scatter(cat1['ra'][mask1], cat1['dec'][mask1], c=cat1[band+'_sky_corr'][mask1], vmin=-vrange, vmax=vrange, s=ms, alpha=alpha, cmap='seismic')
    ax.axis([ramax, ramin, decmin, decmax])
    ax.grid(alpha=0.5)
    ax.set_xlabel('ra')
    ax.set_ylabel('dec')
    # fig.colorbar(im)
    ax.set_title('[dr8] '+band+'-band offset-corrected sky residuals [not masked] (color range: +-{})'.format(vrange))
    plt.savefig('plots/dr8_'+band+'_sky_residuals_region1.png')
    plt.show()
```
